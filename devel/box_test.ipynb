```python

analogOutDict = {
        0: "12,A",
        1: "12,B",
        2: "12,C",
        3: "12,D",
        4: "13,A",
        5: "13,B",
        6: "13,C",
        7: "13,D",
    }
def convert_analog_out(value,vrednost):
    tempList =[]
    if type(value) == int:
        print('int')
        setAnalog = analogOutDict[value].split(",")
        #self.ser.write(f'{SET_DAC},{chip_output},{chip_value}'.encode())
        print(setAnalog[0],setAnalog[1])
    
    if type(value) == list:
        print('list')
        for number in value:
            tempList.append(analogOutDict[number])
            tempList.append(vrednost)
           
          
        print(tempList,vrednost)
        listToStr = ','.join(map(str, tempList))
 
        print(listToStr)
           
```
